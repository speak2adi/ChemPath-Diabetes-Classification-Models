# OVERALL FINDINGS

The comparison shows that no single model performed best across every evaluation metric.

AdaBoost achieved the highest overall accuracy at **78%** and tied with the Decision Tree for the highest recall of the diabetic class at **73%**. However, Gaussian Naive Bayes achieved the highest ROC-AUC score of **0.83**, indicating the strongest overall ability to distinguish between diabetic and non-diabetic patients across different classification thresholds.

Gradient Boosting, Linear SVM, and XGBoost also demonstrated strong discrimination, each achieving a ROC-AUC score of **0.82**.

Random Forest produced the lowest accuracy among the evaluated models at **72%**, while the RBF SVM also performed relatively poorly compared with the Linear SVM.

These results demonstrate that increasing model complexity does not automatically result in better predictive performance. For example, XGBoost achieved an accuracy of **76%**, while the simpler Gradient Boosting model achieved **77%**.

# MEDICAL PERSPECTIVE

Because the objective is to identify patients who may have diabetes, recall for the diabetic class is particularly important.

A false negative occurs when a patient who is actually diabetic is classified as non-diabetic. In a medical screening context, this can be more concerning than a false positive because the patient may not be identified for further investigation.

Based on diabetic-class recall:

- Decision Tree: **0.73**
- AdaBoost: **0.73**
- Gaussian Naive Bayes: **0.71**
- Gradient Boosting: **0.71**

AdaBoost therefore provides a strong balance between identifying diabetic patients and maintaining overall predictive performance.

However, model predictions should not be interpreted as a medical diagnosis. The models in this project are predictive tools evaluated on a specific dataset and would require further validation, calibration, clinical assessment, and external testing before any real-world medical application.

# FINAL CONCLUSION

This project evaluated twelve classification approaches for predicting diabetes using the Pima Indians Diabetes Dataset, ranging from traditional machine learning algorithms to ensemble and boosting methods.

The results showed that model performance varied depending on the evaluation metric used.

AdaBoost achieved the highest overall accuracy at **78%** and tied with the Decision Tree for the highest diabetic-class recall at **73%**. However, Gaussian Naive Bayes achieved the highest ROC-AUC score of **0.83**, while Linear SVM, Gradient Boosting, and XGBoost each achieved **0.82**.

From a medical classification perspective, diabetic-class recall was given particular attention because false negatives represent actual diabetic patients who were incorrectly classified as non-diabetic. Based on this criterion, AdaBoost provided one of the strongest overall balances between identifying diabetic cases and maintaining overall predictive performance.

The results also demonstrated that more complex models do not automatically outperform simpler algorithms. XGBoost, despite being a highly optimized gradient boosting implementation, achieved an accuracy of **76%**, while Gradient Boosting achieved **77%** and Gaussian Naive Bayes achieved **77%** accuracy with a higher ROC-AUC of **0.83**.

Overall, the experiment highlights the importance of evaluating multiple models using several metrics rather than selecting a model based solely on accuracy. The most appropriate model depends on the objective of the application and the relative cost of different types of classification errors.

For this dataset and the selected model configurations, **AdaBoost is a strong candidate when prioritising the identification of diabetic cases, while Gaussian Naive Bayes demonstrates the strongest overall discrimination based on ROC-AUC**.

These findings are specific to the dataset, preprocessing decisions, model configurations, and test split used in this project and should not be interpreted as evidence of clinical effectiveness.
